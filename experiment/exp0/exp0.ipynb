### Data Info

Here is a concise data descriptive summary for the dataset at /Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2:

- Location
  - <mcfolder name="train2" path="/Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2/"></mcfolder>

- Subfolders and contents
  - <mcfolder name="pdbs" path="/Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2/pdbs/"></mcfolder>
    - Raw PDB files extracted to contain single RNA molecules (typically one chain or segment per file).
    - Filenames generally follow a pattern like pdb<id>_<chain>.pdb (e.g., pdb6em4_1.pdb).
    - Note: Some entries may only include sugar–phosphate backbone atoms (e.g., P, OP1/OP2, O5'/C5'/C4'/O4'/C3'/O3'/C2'/O2'/C1') and lack base heavy atoms (N1/N9, O2/O4/O6, N2/N3/N4). For such files, base-pair detection based on base atoms will yield zero, and C1'-C1' distance-based methods are preferred.

  - <mcfolder name="coords" path="/Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2/coords/"></mcfolder>
    - NPY files containing PDB-derived “7-atoms” coordinates per residue.
    - Typical data shape: (L, 7, 3), where L is the number of residues; each entry stores x/y/z in Angstroms for a fixed 7-atom set per nucleotide.
    - The atom ordering is consistent within the pipeline; if a residue is missing one of the expected atoms, the array may include placeholders (e.g., NaN) depending on upstream processing.

  - <mcfolder name="seqs" path="/Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2/seqs/"></mcfolder>
    - FASTA files of sequences derived from the PDB structures.
    - Headers typically encode the PDB identifier and chain; sequences use RNA alphabet (A, U, G, C), with some modified nucleotides normalized to canonical bases (e.g., DT→U) to ensure downstream consistency.

  - <mcfolder name="matrices" path="/Users/xiaojuzhang/Dev/pdb_prune/experiment/exp0/competition/train2/matrices/"></mcfolder>
    - Text files containing base-pair adjacency matrices derived from PDB coordinates.
    - Each matrix is a square, symmetric 0/1 matrix of size L×L, where L is the number of residues in the corresponding RNA.
    - A value of 1 indicates a detected base pair between residue i and residue j; 0 indicates no pair.
    - Filenames generally follow the source PDB and chain naming convention (e.g., pdb1a9n_Q.txt, pdb2wj8_a.txt).
    - Matrices are generated by the base-pair detection script at <mcfile name="pdb_3dto2d_basepair.py" path="/Users/xiaojuzhang/Dev/pdb_prune/utils/pdb_3dto2d_basepair.py"></mcfile>, which supports multiple distance criteria (C1'-C1', N1/N9, or both) and optional coplanarity/H-bond checks.

- Practical notes
  - If a PDB lacks base heavy atoms (as noted in your exp0 notebook for pdb6em4_1.pdb), prefer using the C1'-C1' distance method to build useful matrices; strictly base-atom methods will produce no pairs in such cases.
  - Ensure that the sequence (FASTA) and matrix dimensions align for downstream analyses; mismatches usually indicate truncated structures or missing residues.
  - For large or low-resolution structures (e.g., cryo-EM-derived), consider relaxed distance thresholds if matrices appear overly sparse, and document the criterion used for reproducibility.
